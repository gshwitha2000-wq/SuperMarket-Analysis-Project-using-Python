```python
data['total']=data['Unit price']*data['Quantity']
total_sales=data['total'].sum()
print(total_sales)
```

```python
product_sales=data.groupby('Product line')['total'].sum()
print(product_sales)
```

```python
product_sales.plot(kind='bar')
plt.xlabel('Product line')
plt.ylabel('Total Sales')
plt.title('Product Wise Sales')
plt.show()
```

```python
payment_count=data['Payment'].value_counts()
print(payment_count)
```

```python
plt.pie(payment_count,labels=['Ewallet','Cash','Credit card'],autopct="%1.1f%%")
plt.title("Payment Method")
plt.show()
```

```python
gender_count=data['Gender'].value_counts()
print(gender_count)
```

```python
gender_sales=data.groupby('Gender')['total'].sum()
print(gender_sales)
```

```python
plt.hist(data['gross income'],bins=20)
plt.xlabel("Gross Income")
plt.ylabel("Frequency")
plt.title("Gross Income Distribution")
plt.show()
```

```python
# Plot distribution of customer ratings
plt.figure(figsize=(8, 6))
plt.hist(data['Rating'], bins=10)
plt.title('Customer Rating Distribution')
plt.show()
```
